# Word Count in Text Starting with 'V'

```python
from collections import Counter
import matplotlib.pyplot as plt

# Read the content of the file
def rvv_trace_histogram(filename, title):
    with open(filename, 'r') as file:
        text = file.read()

    # Split the text into words
    words = text.split()

    # Filter words starting with lowercase 'v' and containing '.'
    v_dot_words = [word for word in words if (word.startswith('v') and '.' in word) or word.startswith('vset')]

    # Create a histogram
    # Count the frequency of each word
    word_counts = Counter(v_dot_words)

    # Sort the words by frequency in decreasing order
    sorted_word_counts = dict(sorted(word_counts.items(), key=lambda item: item[1], reverse=True))

    # Create a histogram with sorted bins
    plt.bar(sorted_word_counts.keys(), sorted_word_counts.values(), edgecolor='black')
    plt.xlabel('RVV Instruction')
    plt.ylabel('Count')
    plt.title(title)
    plt.xticks(rotation=90)
    plt.show()

# Read the content of the file
def rvv_trace_proportions(filename, title, cutoff=20):
    with open(filename, 'r') as file:
        text = file.read()

    # Split the text into words
    words = text.split()

    # Filter words starting with lowercase 'v' and containing '.'
    v_dot_words = [word for word in words if (word.startswith('v') and '.' in word) or word.startswith('vset')]

    # Create a histogram
    # Count the frequency of each word
    word_counts = Counter(v_dot_words)

    # Sort the words by frequency in decreasing order
    sorted_word_counts = dict(sorted(word_counts.items(), key=lambda item: item[1], reverse=True))

    # Normalize the counts by dividing by the total number of words
    total_words = sum(sorted_word_counts.values())
    normalized_counts = {word: count / total_words for word, count in sorted_word_counts.items()}

    # Apply cutoff to display only the top 'cutoff' bins
    top_normalized_counts = dict(list(normalized_counts.items())[:cutoff])

    # Create a histogram
    plt.figure(figsize=(cutoff/3, 5))
    plt.bar(top_normalized_counts.keys(), top_normalized_counts.values(), edgecolor='black')
    plt.xlabel('RVV Instruction')
    plt.ylabel('Proportion')
    plt.title(title)
    plt.xticks(rotation=90)
    plt.show()
```

```python
filename = '../../bin/apps/sconv_rvv/bin/riscv-64-linux-rvv-vector_bits_256/sconv-run.trace'
title = 'RVV Instruction Trace Count - Separable Convolution'
rvv_trace_proportions(filename, title)
```

```python
filename = '../../bin/apps/camera_pipe/bin/riscv-64-linux-rvv-vector_bits_256/camera-pipe-run.S'
title = 'RVV Instruction Count - Camera Pipe'
rvv_trace_proportions(filename, title, 30)
```

```python
filename = '../../bin/apps/lens_blur/bin/riscv-64-linux-rvv-vector_bits_256/lens-blur-run.S'
title = 'RVV Instruction Count - Lens Blur'
rvv_trace_proportions(filename, title, 30)
```

```python
import matplotlib.pyplot as plt

# Read the content of the file
def rvv_trace_histogram(filename):
    with open(filename, 'r') as file:
        text = file.read()

    # Split the text into words
    words = text.split()

    # Filter words starting with lowercase 'v' and containing '.'
    v_dot_words = [word for word in words if (word.startswith('v') and '.' in word) or word.startswith('vset')]

    # Create a histogram
    plt.hist(v_dot_words, bins=len(set(v_dot_words)), edgecolor='black')
    plt.xlabel('Words')
    plt.ylabel('Frequency')
    plt.title('Histogram of words starting with "v" and containing "."')
    plt.xticks(rotation=90)
    plt.show()
```
