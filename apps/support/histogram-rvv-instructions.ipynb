# Word Count in Text Starting with 'V'

```python
import matplotlib.pyplot as plt

# Read the content of the file
def rvv_trace_histogram(filename):
    with open(filename, 'r') as file:
        text = file.read()

    # Split the text into words
    words = text.split()

    # Filter words starting with lowercase 'v' and containing '.'
    v_dot_words = [word for word in words if (word.startswith('v') and '.' in word) or word.startswith('vset')]

    # Create a histogram
    plt.hist(v_dot_words, bins=len(set(v_dot_words)), edgecolor='black')
    plt.xlabel('Words')
    plt.ylabel('Frequency')
    plt.title('Histogram of words starting with "v" and containing "."')
    plt.xticks(rotation=90)
    plt.show()
```

```python
filename = '../../bin/apps/sconv_rvv/bin/riscv-64-linux-rvv-vector_bits_512/process-riscv.trace'
rvv_trace_histogram(filename)
```
